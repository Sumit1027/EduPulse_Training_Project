EXPLORATORY DATA ANALYSIS

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../Dataset/StudentPerformanceFactors_Cleaned.csv")
```

```python
df.head()
```

```python
# Histogram :
plt.figure(figsize=(8,5))
sns.histplot(df["Exam_Score"], bins=15, kde=True)
plt.title("Distribution Of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number Of Students")
plt.show()
```

```python
# Box Plot:
plt.figure(figsize=(8,5))
sns.boxplot(x=df["Exam_Score"], color="pink")
plt.title("Box Plot Of Exam Scores")
plt.xlabel("Exam Score")
plt.show()
# There are many Outliers dots.
```

```python
# Count Plot:
plt.figure(figsize=(6,4))
sns.countplot(x="Gender", data=df, palette="Set2")
plt.title("Distribution Of Students By Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(x="Parental_Involvement", data=df, color="lightgreen")
plt.title("Distribution Of Parental Involvement")
plt.xlabel("Parental Involvement")
plt.ylabel("Number Of Students")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(x="Teacher_Quality", data=df, color="skyblue")
plt.title("Distribution Of Teacher Quality")
plt.xlabel("Teacher Quality")
plt.ylabel("Number Of Students")
plt.show()
```

CORRELATION HEATMAP

```python
plt.figure(figsize=(10,8))
correlation = df.select_dtypes(include="number").corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap Of Numerical Features")
plt.show()
```

SCATTER PLOT

```python
plt.figure(figsize=(8,5))
sns.scatterplot( data=df, x="Hours_Studied", y="Exam_Score", hue="Gender", alpha=0.7)
plt.title("Hours Studied Vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

BOX PLOT BETWEEN TEACHER QUALITY AND EXAM SCORE:

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Teacher_Quality", y="Exam_Score", data=df, palette="coolwarm")
plt.title("Teacher Quality Vs Exam Score ")
plt.xlabel("Teacher Quality")
plt.ylabel("Exam Score")
plt.show()
```

RELATIONSHIP BETWEEN PARENTAL INVOLVEMENT AND EXAM SCORE

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Parental_Involvement", y="Exam_Score", data=df, palette="coolwarm")
plt.title("Parental Involvement Vs Exam Score ")
plt.xlabel("Parental Involvement")
plt.ylabel("Exam Score")
plt.show()
```

RELATIONSHIP BETWEEN INTERNET ACCESS AND EXAM SCORES

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Internet_Access", y="Exam_Score", data=df, palette="coolwarm")
plt.title("Internet Access Vs Exam Score ")
plt.xlabel("Internet Access")
plt.ylabel("Exam Score")
plt.show()
```
