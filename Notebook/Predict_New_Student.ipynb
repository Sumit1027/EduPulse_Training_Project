```python
import pandas as pd
import joblib
from sklearn.preprocessing import LabelEncoder
```

LOAD THE SAVED MODEL AND SCALER

```python
model = joblib.load("../Model/final_linear_regression_model.pkl")
scaler = joblib.load("../Model/scaler.pkl")
print("Model And Scaler Loaded Successfully.")
```

```python
df = pd.read_csv("../Dataset/StudentPerformanceFactors_Cleaned.csv")
print("DataSet Are Loaded Successfully.")
```

ENCODE CATEGORICAL FEATURES

```python
label_encoder = LabelEncoder()
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column] = label_encoder.fit_transform(df[column])
```

CHECK FEATURE ORDER

```python
X = df.drop("Exam_Score", axis=1)
print("Feature Order:")
print(X.columns.tolist())
```

CREATE A NEW STUDENT RECORD

```python
new_student = pd.DataFrame({
    "Hours_Studied": [8],
    "Attendance": [90],
    "Parental_Involvement": [2],
    "Access_to_Resources": [2],
    "Extracurricular_Activities": [1],
    "Sleep_Hours": [7],
    "Previous_Scores": [82],
    "Motivation_Level": [2],
    "Internet_Access": [1],
    "Tutoring_Sessions": [3],
    "Family_Income": [1],
    "Teacher_Quality": [2],
    "School_Type": [1],
    "Peer_Influence": [2],
    "Physical_Activity": [4],
    "Learning_Disabilities": [0],
    "Parental_Education_Level": [2],
    "Distance_from_Home": [1],
    "Gender": [1]
})

new_student
```

SCALE THE NEW STUDENT DATA

```python
new_student_scaled = scaler.transform(new_student)
print("New Student Data Scaled Successfully.")
```

PREDICT THE EXAM SCORE

```python
predicted_score = model.predict(new_student_scaled)
print("=" * 50)
print("Exam Score Prediction")
print("=" * 50)
print(f"Predicted Exam Score: {predicted_score[0]:.2f}")
print("=" * 50)
```
